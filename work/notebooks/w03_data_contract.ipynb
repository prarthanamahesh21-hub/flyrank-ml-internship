```python
import duckdb

con = duckdb.connect()

con.register("dim_content", content._data.table)
con.register("fact_content_daily_performance", performance._data.table)
con.register("fact_content_query_90d", query_90d._data.table)

print("DuckDB setup complete")
```

```python
grain_check = con.sql("""
SELECT
    COUNT(*) AS march_rows,
    COUNT(DISTINCT content_hash_id || '|' || CAST(report_date AS VARCHAR)) AS unique_page_day_pairs,
    COUNT(*) -
        COUNT(DISTINCT content_hash_id || '|' || CAST(report_date AS VARCHAR))
        AS duplicate_page_day_rows
FROM fact_content_daily_performance
WHERE report_date >= DATE '2026-03-01'
  AND report_date < DATE '2026-04-01'
""").df()

grain_check
```

```python
march_window = con.sql("""
SELECT
    COUNT(*) AS march_rows,
    MIN(report_date) AS first_report_date,
    MAX(report_date) AS last_report_date
FROM fact_content_daily_performance
WHERE report_date >= DATE '2026-03-01'
  AND report_date < DATE '2026-04-01'
""").df()

march_window
```

```python
availability_check = con.sql("""
SELECT
    COUNT(*) AS march_rows,
    COUNT(*) FILTER (WHERE gsc_data_available IS TRUE) AS gsc_available_rows,
    COUNT(*) FILTER (WHERE gsc_data_available IS NOT TRUE) AS gsc_not_available_rows
FROM fact_content_daily_performance
WHERE report_date >= DATE '2026-03-01'
  AND report_date < DATE '2026-04-01'
""").df()

availability_check
```

## 2. Fields: feature / label / context / excluded

*Sort every field you plan to touch into these four buckets. Excluded needs a why.*

Feature: content_age_days, word_count, impressions_90d, clicks_90d, and pageviews_90d. These are historical or page-level signals that can be known when the page is being considered for review.

Label: A future-performance outcome used only to evaluate whether a page should have been prioritized. It is not used as an input feature.

Context: content_id, client_id, content_type, and other identifying/context fields needed to group or describe pages.

Excluded: Any future-performance field or field calculated from the outcome window is excluded from the feature set because it would not be known at the decision moment and would cause leakage.

```python
features_march = con.sql("""
WITH performance_90d AS (
    SELECT
        content_hash_id,

        SUM(
            CASE
                WHEN gsc_data_available IS TRUE
                THEN COALESCE(gsc_impressions, 0)
                ELSE 0
            END
        ) AS gsc_impressions_90d,

        SUM(
            CASE
                WHEN gsc_data_available IS TRUE
                THEN COALESCE(gsc_clicks, 0)
                ELSE 0
            END
        ) AS gsc_clicks_90d,

        SUM(
            CASE
                WHEN ga4_data_available IS TRUE
                THEN COALESCE(ga4_pageviews, 0)
                ELSE 0
            END
        ) AS ga4_pageviews_90d

    FROM fact_content_daily_performance

    WHERE report_date >= DATE '2025-12-01'
      AND report_date < DATE '2026-03-01'

    GROUP BY content_hash_id
)

SELECT
    c.content_hash_id,

    DATE_DIFF(
        'day',
        CAST(c.content_created_date AS DATE),
        DATE '2026-03-01'
    ) AS content_age_days,

    c.word_count,
    p.gsc_impressions_90d,
    p.gsc_clicks_90d,
    p.ga4_pageviews_90d

FROM dim_content c

LEFT JOIN performance_90d p
    ON c.content_hash_id = p.content_hash_id

WHERE c.content_created_date < DATE '2026-03-01'
  AND c.is_published IS TRUE
  AND c.is_deleted IS FALSE
""").df()

print("Feature frame shape:", features_march.shape)
features_march.head()
```

```python
# Future outcome: April 2026 GSC clicks

future_label = con.sql("""
SELECT
    content_hash_id,
    SUM(
        CASE
            WHEN gsc_data_available IS TRUE
            THEN COALESCE(gsc_clicks, 0)
            ELSE 0
        END
    ) AS future_gsc_clicks
FROM fact_content_daily_performance
WHERE report_date >= DATE '2026-04-01'
  AND report_date < DATE '2026-05-01'
GROUP BY content_hash_id
""").df()

future_label["label"] = (
    future_label["future_gsc_clicks"] > 0
).astype(int)

model_df = features_march.merge(
    future_label[["content_hash_id", "label"]],
    on="content_hash_id",
    how="inner"
)

print("Model frame shape:", model_df.shape)
print("Positive labels:", model_df["label"].sum())
print("Negative labels:", (model_df["label"] == 0).sum())
```
