```python
# Build a leakage-safe Week 4 feature frame
# using the existing 90-day query table.

features_w04 = con.sql("""
WITH query_by_content AS (
    SELECT
        content_hash_id,
        SUM(COALESCE(impressions_90d, 0)) AS gsc_impressions_90d,
        SUM(COALESCE(clicks_90d, 0)) AS gsc_clicks_90d,
        SUM(COALESCE(content_visible_query_count, 0)) AS visible_query_count
    FROM fact_content_query_90d
    GROUP BY content_hash_id
)

SELECT
    c.content_hash_id,

    DATE_DIFF(
        'day',
        CAST(c.content_created_date AS DATE),
        DATE '2026-03-01'
    ) AS content_age_days,

    c.word_count,

    COALESCE(q.gsc_impressions_90d, 0) AS gsc_impressions_90d,
    COALESCE(q.gsc_clicks_90d, 0) AS gsc_clicks_90d,
    COALESCE(q.visible_query_count, 0) AS visible_query_count

FROM dim_content c

LEFT JOIN query_by_content q
    ON c.content_hash_id = q.content_hash_id

WHERE c.content_created_date < DATE '2026-03-01'
  AND c.is_published IS TRUE
  AND c.is_deleted IS FALSE
""").df()

print("Feature frame shape:", features_w04.shape)
features_w04.head()
```

## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*


**Rule:** Prioritize pages for a possible content refresh when they are older and have meaningful search visibility. Content age is used as a staleness signal, while 90-day Google Search Console impressions are used as a search-opportunity signal.

**Signal check verdicts:**

* **`content_age_days` — MIXED:** The age bucket check did not show higher median search activity for older pages. Median impressions and clicks were zero across all age buckets, so age alone does not demonstrate higher search activity.
* **`gsc_impressions_90d` — CONFIRMED:** The 1,000+ impression bucket had a median of 4 clicks, while the lower-impression buckets had a median of 0 clicks. This supports using impressions as a directional measure of search visibility.

**Reason code:**

* `STALE_HIGH_VALUE` — The page is relatively old and has meaningful search visibility, so it receives higher priority for refresh review.

```python
# Build the baseline score

baseline = features_w04.copy()

# Age points
baseline["age_points"] = pd.cut(
    baseline["content_age_days"],
    bins=[-1, 90, 180, 365, np.inf],
    labels=[0, 1, 2, 3]
).astype(int)

# Search visibility points
baseline["impression_points"] = pd.cut(
    baseline["gsc_impressions_90d"],
    bins=[-1, 0, 10, 100, 1000, np.inf],
    labels=[0, 1, 2, 3, 4]
).astype(int)

# Final score
baseline["score"] = (
    baseline["age_points"] +
    baseline["impression_points"]
)

# One reason code
baseline["reason_code"] = "STALE_HIGH_VALUE"

# Action label
baseline["action"] = np.where(
    baseline["score"] >= 5,
    "Refresh",
    "Monitor"
)

# Rank pages
baseline = baseline.sort_values(
    ["score", "content_age_days", "gsc_impressions_90d"],
    ascending=[False, False, False]
).reset_index(drop=True)

baseline["rank"] = np.arange(1, len(baseline) + 1)

# Final output columns
baseline_output = baseline[
    [
        "rank",
        "content_hash_id",
        "score",
        "reason_code",
        "action",
        "content_age_days",
        "gsc_impressions_90d",
        "gsc_clicks_90d"
    ]
]

print("Ranked pages:", len(baseline_output))
print("\nTop 10:")
print(baseline_output.head(10).to_string(index=False))
```
