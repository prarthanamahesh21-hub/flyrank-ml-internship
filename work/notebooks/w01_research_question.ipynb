## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### The question

**Search question:**
Which content pages should be reviewed first for a potential refresh, based on their observable search performance, content age, engagement, and visibility signals?

**Decision:**
Decide which pages should receive editorial or SEO review first.

**Unit of analysis:**
One content page.

**Output:**
A ranked list of content pages with an opportunity or priority score.

**Action:**
An editor or SEO reviewer can inspect the highest-ranked pages and decide whether a content refresh or further investigation is appropriate.

**Cost of a wrong recommendation:**
A false positive could waste limited editorial effort on a page that is not a priority. A false negative could cause an important page with a potential improvement opportunity to be overlooked.

The goal is therefore to support prioritization, not to automatically decide that a page must be refreshed.

### Quick look at the data

The starter dataset contains **30,000 content pages** across **32 unique clients** and **44 columns** of page-level information. The median content age is **236 days**, and **17,728 pages are older than 180 days**.

These numbers make the refresh-prioritization problem worth investigating because the dataset contains a large number of pages and a substantial portion of them are relatively old. A ranked review queue could help an editor or SEO team decide which pages to investigate first instead of manually reviewing every page. However, content age alone does not mean that a page needs a refresh, so other search and engagement signals will also need to be considered.


```python
import pandas as pd

url = "https://raw.githubusercontent.com/prarthanamahesh21-hub/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Number of pages:", len(df))
print("Number of columns:", len(df.columns))
print("Number of unique clients:", df["client_id"].nunique())
```

```python
print("Median content age (days):", df["content_age_days"].median())
print("Pages older than 180 days:", (df["content_age_days"] > 180).sum())
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

### What I can and cannot claim

**What I can claim:**
The dataset provides page-level search, engagement, content-age, and other observable signals that can be used to investigate whether pages can be prioritized for review. The size and structure of the dataset make it reasonable to explore a ranking approach for editorial review.

**What I cannot claim yet:**
I cannot claim that a particular page definitely needs a refresh, that a high opportunity score will cause better search performance, or that an ML model will outperform a simple rule until those claims are tested. A ranking should be treated as decision support for further review rather than as proof that a refresh will improve a page's performance.
